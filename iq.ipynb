#### 1. What is Python and why is it popular in Data Science?

Python is a high-level, interpreted programming language. It is popular in Data Science because it is easy to learn, has powerful libraries like NumPy, Pandas, Matplotlib, Scikit-learn, and supports fast data analysis and machine learning.

#### 2. Difference between List, Tuple, Set, and Dictionary

List: Ordered, mutable, allows duplicates

Tuple: Ordered, immutable, allows duplicates

Set: Unordered, mutable, does not allow duplicates

Dictionary: Stores data as key-value pairs, mutable, keys are unique

#### 3. What is a Lambda function?

A lambda function is an anonymous function written in a single line, used for short operations.

lambda x: x*x

#### 4. What is map(), filter(), reduce()?

map(): Applies a function to all elements

filter(): Filters elements based on a condition

reduce(): Reduces elements to a single value (from functools)

#### 5. Difference between NumPy array and Python list?

NumPy arrays are faster and memory efficient

They support vectorized operations

Lists can store different data types, NumPy arrays usually store same type

#### 6. What is Pandas used for?

Pandas is used for data manipulation and analysis, such as data cleaning, handling missing values, filtering, grouping, and working with datasets.

#### 7. How do you handle missing values?

Remove missing values (drop)

Fill with mean, median, or mode

Use forward fill or backward fill

Use model-based imputation

#### 8. Difference between loc and iloc

loc: Label-based indexing

iloc: Integer-based indexing

#### 9. What is Mean, Median, Mode?

Mean: Average of values

Median: Middle value

Mode: Most frequent value

#### 10. What is Variance and Standard Deviation?

Variance measures data spread from the mean

Standard Deviation is the square root of variance and shows how much values deviate

#### 11. What is Correlation?

Correlation measures the relationship between two variables, ranging from -1 to +1.

#### 12. What is Machine Learning?

Machine Learning is a subset of AI where systems learn from data and improve performance without being explicitly programmed.

#### 13. Types of Machine Learning

Supervised Learning

Unsupervised Learning

Semi-Supervised Learning

Reinforcement Learning

#### 14. Difference between Regression and Classification

Regression predicts continuous values

Classification predicts categorical labels

#### 15. What is Linear Regression?

Linear Regression models the relationship between input and output using a straight line.

#### 16. Assumptions of Linear Regression

Linearity

Independence

Homoscedasticity

Normality

No multicollinearity

#### 17. What is R² score?

R² score measures how well the model explains variance in the target variable. Values range from 0 to 1.

#### 18. What is Logistic Regression?

Logistic Regression is a classification algorithm used to predict binary outcomes using probabilities.

#### 19. Why sigmoid function is used?

Sigmoid converts output into a probability between 0 and 1, making it suitable for classification.

#### 20. Difference between Linear and Logistic Regression

Linear Regression predicts continuous values

Logistic Regression predicts probabilities and class labels

#### 21. What is Margin in SVM?

Margin is the distance between the decision boundary and the nearest data points.

#### 22. What are Support Vectors?

Support vectors are the data points closest to the hyperplane that influence the model.

#### 23. What is Kernel Trick?

Kernel trick allows SVM to transform data into higher dimensions without explicit computation.

#### 24. Types of Kernels

Linear

Polynomial

RBF (Gaussian)

Sigmoid

#### 25. What is C parameter in SVM?

C controls the trade-off between margin width and classification error.

#### 26. What is Train-Test Split?

Train-Test split divides data into training and testing sets to evaluate model performance.

#### 27. Difference between Batch and Streaming

Batch processing handles large data at once

Streaming processing handles data in real time

#### 28. What is ETL?

ETL stands for Extract, Transform, Load, used in data pipelines to prepare data for analysis.

#### 29. Difference between Bias and Variance

Bias: Error due to oversimplified model

Variance: Error due to overly complex model

A good model balances both.
